### 문제 1. 두 수 중 작은 값

두 정수를 입력받아 더 작은 값을 아래 형식으로 출력하시오. 두 수가 같은 경우에는 그 값을 출력한다.

**입력**  
한 줄에 두 정수가 공백으로 주어진다.

**출력**  
더 작은 값을 아래 출력 예와 같은 문장 형식으로 출력한다.

**입력 예**
```text
13 8
```

**출력 예**
```text
두 수 중 작은 값은 8입니다.
```

````python
# 여기에 코드를 작성하세요.
n1, n2 = map(int, input().split())
if n1 > n2:
  print(f"두 수 중 작은 값은 {n2}입니다.")
if n1 < n2:
  print(f"두 수 중 작은 값은 {n1}입니다.")
elif n1 == n2:
  print(n1)
````

### 문제 4. 조건을 만족하는 수 분석

**입력**  
정수 N

**출력**  
첫 줄에 조건을 만족하는 수의 개수, 둘째 줄에 그 수들의 합

**입력 예**
```text
20
```

**출력 예**
```text
3
36
```

````python
# 여기에 코드를 작성하세요.
n = int(input())
s = 0
x = 0
for i in range(0, n):
  if i == 0:
    continue
  if i % 3 == 0 and i % 2 == 0:
    s = s + i
    x = x + 1
print(x)
print(s)


````

정수를 한 줄에 하나씩 입력받는다. 음수가 입력되면 입력을 종료하고, 그 전에 입력된 정수의 개수와 합을 출력하시오. 종료에 사용된 음수는 개수와 합에 포함하지 않는다.

**입력**  
음수가 나올 때까지 한 줄에 하나씩 정수

**출력**  
첫 줄에 음수 이전까지 입력된 정수의 개수, 둘째 줄에 그 정수들의 합

**입력 예**
```text
5
8
0
3
-1
```

**출력 예**
```text
4
16
```

````python
# 여기에 코드를 작성하세요.
n = int(input())
sum = 0
count = 0
while True:
  if n < 0:
    print(count)
    print(sum)
    break
  else :
    sum =  sum + n
    count = count + 1
    n=int(input())
````

### 문제 6. 양수·음수·0의 개수

정수 N개를 입력받아 양수, 음수, 0이 각각 몇 개인지 출력하시오.

**입력**  
첫 줄 N, 둘째 줄 정수 N개

**출력**  
양수, 음수, 0의 개수를 아래 출력 예와 같은 형식으로 한 줄에 출력한다.

**입력 예**
```text
7
-2 5 0 7 -1 0 3
```

**출력 예**
```text
양수: 3개, 음수: 2개, 0: 2개
```

````python
# 여기에 코드를 작성하세요.
n=int(input())
up=0
lo=0
ze=0
num=list(map(int, input().split()))
for i in range(0, len(num)):
  if num[i] == 0:
    ze = ze + 1
  if num[i] > 0:
    up = up + 1
  if num[i] < 0:
    lo = lo + 1
print(f"양수: {up}개, 음수: {lo}개, 0: {ze}개")
````

**입력**  
첫 줄 출석 번호가 불린 횟수 N, 둘째 줄 N개의 출석 번호, 셋째 줄 확인할 출석 번호 X

```text
8
3 5 2 3 7 3 1 3

```

**입력**  
첫 줄 N, 둘째 줄 학생 점수 N개

**출력**  
첫 줄에는 내림차순으로 정렬된 전체 점수를 공백으로 구분하여 출력한다. 둘째 줄과 셋째 줄에는 각각 가장 높은 점수와 가장 낮은 점수를 출력 예와 같은 형식으로 출력한다.

**입력 예**
```text
5
72 55 90 68 81
```

**출력 예**
```text
90점 81점 72점 68점 55점

최저 점수: 55
```

````python
# 여기에 코드를 작성하세요.
n=int(input())
num=list(map(int, input().split()))
num.sort(reverse = True)
for i in range(0, len(num)):
  print(f"{num[i]}점", end=" ")
print()
print(f"최고점수: {num[0]}")
print(f"최저점수: {num[len(num)-1]}")
````

### 문제 13. 특정 문자의 모든 위치

문자열과 찾을 문자 하나를 입력받아 해당 문자가 있는 모든 위치(index)를 출력하시오. 첫 줄에는 모든 위치를 공백으로 구분하여 출력하고, 둘째 줄에는 해당 문자가 나타난 총 횟수를 출력하시오. index는 0부터 시작하며, 찾을 문자는 한 번 이상 존재한다.

**입력**  
첫 줄 문자열, 둘째 줄 찾을 문자

**출력**  
첫 줄에 모든 위치(공백 구분), 둘째 줄에 총 횟수

**입력 예**
```text
strawberry
r
```

**출력 예**
```text
위치: 2번 7번 8번
총 횟수: 3회
```

````python
# 여기에 코드를 작성하세요.
text=input()
ftext=input()
to=0
re=-1
li=[]
lj=[]
for w in text:
  lj.append(w)
for ch in range(0, len(lj)):
  re = re + 1
  if lj[ch] == ftext:
    to = to + 1
    li.append(re)

print("위치:", end=" ")
for i in range(0, len(li)):
  print(f"{li[i]}번", end = " ")
print()
print(f"총 횟수: {to}회")
````

### 문제 15. 문자 종류 분석

영문 대문자·소문자·숫자가 섞인 문자열을 입력받아 각각의 개수를 출력하시오.

**입력**  
영문 대문자·소문자·숫자로 이루어진 문자열

**출력**  
대문자, 소문자, 숫자의 개수를 각각 한 줄씩 출력

**입력 예**
```text
AbCde1234
```

**출력 예**
```text
대문자: 2
소문자: 3
숫자: 4
```

````python
# 여기에 코드를 작성하세요.
text=input()
up=0
lo=0
nu=0
for ch in text:
  if ch.isupper():
    up = up + 1
  if ch.islower():
    lo = lo + 1
  if ch.isdigit():
    nu = nu + 1
print(f"대문자: {up}")
print(f"소문자: {lo}")
print(f"숫자: {nu}")
````

정수 N개를 입력받아 최솟값과 그 값이 몇 번째인지(1번째부터 시작)를 두 줄로 출력하시오. 최솟값은 입력에서 한 번만 등장한다. 이 문제에서는 min() 과 index() 를 사용하지 않는다.

첫 줄 N, 둘째 줄 정수 N개

**출력**  
첫 줄에 최솟값, 둘째 줄에 그 값의 위치(1번째부터 시작)

**입력 예**
```text
5
18 7 25 3 11
```

**출력 예**
```text
3
4
```

````python
# 여기에 코드를 작성하세요.
n=int(input())
num=list(map(int, input().split()))
m=num[0]
s=0
l=0
for i in range(0, len(num)):
  s = s + 1
  if m > num[i]:
    m=num[i]
    l = s
print(m)
print(l)
````

### 문제 18. 기준 이상 값 분석

첫 줄 N, 둘째 줄 정수 N개, 셋째 줄 X

첫 줄에 조건을 만족하는 값들의 합, 둘째 줄에 개수

**입력 예**
```text
5
70 45 90 60 80
60
```

```text
300
4
```

````python
# 여기에 코드를 작성하세요.
n=int(input())
num=list(map(int, input().split()))
x=int(input())
sum=0
l=0
for i in range(0, len(num)):
  if num[i] >= x:
    sum = sum + num[i]
    l = l + 1
print(sum)
print(l)
````

### 문제 19. 평균보다 작은 점수

학생 점수 N개를 입력받아 평균을 구한 뒤, 평균보다 작은 점수들을 입력 순서대로 출력하고 다음 줄에 그 개수를 출력하시오. 평균 = 전체 점수의 합 ÷ N. 평균보다 작은 점수는 한 개 이상 존재한다.

**입력**  
첫 줄 N, 둘째 줄 학생 점수 N개

**출력**  
첫 줄에 평균보다 작은 점수들(입력 순서, 공백 구분), 둘째 줄에 그 개수

**입력 예**
```text
5
50 70 90 40 100
```

**출력 예**
```text
평균보다 작은 점수: 50점 40점
학생 수: 2명
```

````python
# 여기에 코드를 작성하세요.
n=int(input())
num=list(map(int, input().split()))
all=0
for i in range(0, len(num)):
  all = all + num[i]
aver=0
aver = all / len(num)
s=0
print("평균보다 작은 점수: ",end = " ")
for a in range(0, len(num)):
  if num[a] < aver:
    s = s  + 1
    print(f"{num[a]}점", end = " ")
print()
print(f"학샏 수: {s}명")
````
